    "Kelas" : [2, 3, 4, 5],

}

```python
print(data)
```

```python
df = pd.DataFrame(data)
df
```

```python
print('Menampilkan head \n')
df.head()
```

```python
print('Menampilkan describe \n')
df.describe()
```

```python
print('Menampilkan info \n')
df.info()
```

```python
print(df.loc[0, "Nama"])
print(df.iloc[1, 1])
```
